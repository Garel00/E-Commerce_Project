Análisis de datos con Pandas

```python
import numpy
import pandas as pd
import pyarrow
import huggingface_hub

from huggingface_hub import hf_hub_download
import os

os.makedirs("/content/data/samples", exist_ok=True)

tables = [
    "customers",
    "products",
    "sessions",
    "orders",
    "order_items",
    "returns",
    "support_tickets"
]

for table in tables:
    hf_hub_download(
        repo_id="HariJha/FreeEcommerceTwin14M",
        filename=f"{table}_sample_50rows.csv",
        repo_type="dataset",
        local_dir="/content/data/samples"
    )

print("Descarga de muestras completada.")

path = "/content/data/samples/"

tables = [
    "customers",
    "products",
    "sessions",
    "orders",
    "order_items",
    "returns",
    "support_tickets"
]

for table in tables:

    df = pd.read_csv(path + f"{table}_sample_50rows.csv")

    print("\n" + "=" * 60)
    print(f"TABLA: {table.upper()}")
    print("=" * 60)

    print("\nDimensiones:")
    print(df.shape)

    print("\nColumnas y tipos de datos:")
    print(df.dtypes)

    print("\nPrimeros 5 registros:")
    display(df.head())

    print("\n" + "=" * 60)
    print(f"CALIDAD DE DATOS: {table.upper()}")
    print("=" * 60)

    print("\nValores nulos por columna:")
    print(df.isnull().sum())

    print("\nRegistros duplicados:")
    print(df.duplicated().sum())

    print("\nPorcentaje de valores nulos:")
    print((df.isnull().mean() * 100).round(2))

    print("\n" + "=" * 60)
    print(f"ESTADÍSTICA DESCRIPTIVA: {table.upper()}")
    print("=" * 60)

    display(df.describe(include="all").T)


```

Pyspark

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("EcommerceDataAnalysis") \
    .master("local[*]") \
    .getOrCreate()

print("Spark iniciado correctamente")
print("Versión de Spark:", spark.version)

spark_tables = {}

for table in tables:

    df = spark.read.csv(
        path + f"{table}_sample_50rows.csv",
        header=True,
        inferSchema=True
    )

    spark_tables[table] = df

    print("\n" + "=" * 60)
    print(f"TABLA: {table.upper()}")
    print("=" * 60)

    print("Dimensiones:", df.count(), "filas")

    print("\nEsquema:")
    df.printSchema()

    print("\nPrimeros 5 registros:")
    df.show(5)
```
